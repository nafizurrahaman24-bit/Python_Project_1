How are in-demand skills trending for Data Analysts?

Methology

1. Aggregate skill counts monthly
2. Re-analyze based on percentage of total jobs
3. Plot the monthly skill demand

```python
# Importing Libraries
import ast
import seaborn as sns
import pandas as pd
from datasets import load_dataset 
import matplotlib.pyplot as plt

# Loading Data
dataset = load_dataset('lukebarousse/data_jobs')
df = dataset['train'].to_pandas()

# Data Cleanup
df['job_posted_date'] = pd.to_datetime(df['job_posted_date'])
df['job_skills'] = df['job_skills'].apply( lambda x: ast.literal_eval (x) if pd.notna(x) else (x) )
```

```python
df_DA_US = df[(df['job_title_short'] == 'Data Analyst') & (df['job_country'] == 'United States')].copy()
```

```python
df_DA_US['job_posted_month_no'] = df_DA_US['job_posted_date'].dt.month
```

```python
df_DA_US_explode = df_DA_US.explode('job_skills')
```

```python
df_DA_US_pivot =df_DA_US_explode.pivot_table(
    index='job_posted_month_no',
    columns='job_skills',
    aggfunc='size',
    fill_value=0
)

# Add total row
df_DA_US_pivot.loc['Total'] = df_DA_US_pivot.sum()

# Sort columns by total counts
df_DA_US_pivot = df_DA_US_pivot[
    df_DA_US_pivot.loc['Total']
    .sort_values(ascending=False)
    .index
]

# Remove total row
df_DA_US_pivot = df_DA_US_pivot.drop(index='Total')

df_DA_US_pivot
```

```python
DA_totals = df_DA_US.groupby('job_posted_month_no').size()
```

```python
df_DA_US_percent = df_DA_US_pivot.div(DA_totals/100, axis=0)
```

```python
df_DA_US_percent = df_DA_US_percent.reset_index()
df_DA_US_percent['job_posted_month'] = df_DA_US_percent['job_posted_month_no'].apply(lambda x: pd.to_datetime(x, format='%m').strftime('%b'))
df_DA_US_percent = df_DA_US_percent.set_index('job_posted_month')
df_DA_US_percent = df_DA_US_percent.drop(columns='job_posted_month_no')
```

```python
df_DA_US_percent
```

```python
import matplotlib.pyplot as plt
import matplotlib.ticker as ticker
import seaborn as sns

sns.set_theme(style='ticks')

fig, ax = plt.subplots(figsize=(12, 6))

# Top 5 trending skills
df_plot = df_DA_US_percent.iloc[:, :5]

sns.lineplot(
    data=df_plot,
    dashes=False,
    linewidth=3,
    palette='tab10',
    ax=ax
)

# Remove legend
ax.get_legend().remove()

# Titles and labels
ax.set_title(
    'Trending Top Skills for Data Analysts in the US',
    fontsize=16,
    fontweight='bold'
)

ax.set_ylabel('Likelihood in Job Postings (%)')
ax.set_xlabel('2023')

# Format Y-axis as percentages
ax.yaxis.set_major_formatter(
    ticker.FuncFormatter(lambda y, _: f'{y:.0f}%')
)

# Make room on right side for data labels
ax.set_xlim(-0.2, len(df_plot.index) + 0.8)

# Offsets to prevent overlapping labels
offsets = [0.5, -0.5, 1.0, -1.0, 0]

for i, column in enumerate(df_plot.columns):
    ax.text(
        len(df_plot.index) - 1 + 0.2,
        df_plot.iloc[-1, i] + offsets[i],
        column,
        fontsize=10,
        va='center'
    )

sns.despine()

plt.tight_layout()
plt.show()
```
